```python
import pandas as pd
import matplotlib.pyplot as plt

reddit = pd.read_csv("reddit_comments.csv")
youtube = pd.read_csv("youtube_comments.csv")
```

```python
plt.figure()
reddit['sentiment'].hist()
plt.title("Reddit Sentiment Score Distribution")
plt.xlabel("Sentiment Score")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure()
youtube['sentiment_score'].hist()
plt.title("YouTube Sentiment Score Distribution")
plt.xlabel("Sentiment Score")
plt.ylabel("Count")
plt.show()
```

```python
reddit_topics = (
    reddit['comment_matched_topics']
    .fillna("")
    .str.split("|")
    .explode()
)

reddit_topics = reddit_topics[reddit_topics != ""]

plt.figure()
reddit_topics.value_counts().plot(kind='bar')
plt.title("Reddit Topic Frequency")
plt.xlabel("Topic")
plt.ylabel("Count")
plt.show()
```

```python
yt_topics = (
    youtube['topics']
    .fillna("")
    .str.split("|")
    .explode()
)

yt_topics = yt_topics[yt_topics != ""]

plt.figure()
yt_topics.value_counts().plot(kind='bar')
plt.title("YouTube Topic Frequency")
plt.xlabel("Topic")
plt.ylabel("Count")
plt.show()
```

```python
reddit_barriers = (
    reddit['barriers']
    .fillna("")
    .str.split("|")
    .explode()
)

reddit_barriers = reddit_barriers[reddit_barriers != ""]

reddit_barriers.value_counts().plot(kind="bar")
plt.title("Reddit Barrier Types")
plt.show()
```
